```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3,4,5])
y = np.array([12,22,35,42,54])
plt.plot(x,y,marker = True)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
lrmod = LinearRegression()
lrmod.fit(x.reshape(-1,1),y)
```

```python
print(lrmod.coef_)
print(lrmod.intercept_)
```

```python
pred = lrmod.predict([[5.2]])
print(pred)
```
